```python
import numpy as np
import torch
import torch.nn as nn
import torch.optim as optim
import re

from datasets import load_dataset
from collections import Counter
```

```python
print("Loading IMDB dataset from Hugging face...")
dataset = load_dataset("stanfordnlp/imdb")

num_words = 10000
def tokenize(text):
    return re.findall(r"\w+", text.lower())

print("Building vocabulary...")
word_counts = Counter()

for text in dataset["train"]["text"]:
    word_counts.update(tokenize(text))

vocab = [word for word, count in word_counts.most_common(num_words)]
word_to_index = {word: i for i, word in enumerate(vocab)}

def vectorize_texts(texts, word_to_index, dimension=10000):
    results = np.zeros((len(texts), dimension), dtype=np.float32)
    for i, text in enumerate(texts):
        for word in tokenize(text):
            if word in word_to_index:
                results[i, word_to_index[word]] = 1.0
    return results


print("Vectorizing train and test data...")
X_train = vectorize_texts(dataset["train"]["text"], word_to_index, num_words)
Y_train = np.array(dataset["train"]["label"], dtype=np.float32)

X_test = vectorize_texts(dataset["test"]["text"], word_to_index, num_words)
Y_test = np.array(dataset["test"]["label"], dtype=np.float32)

X_train_t = torch.tensor(X_train)
Y_train_t = torch.tensor(Y_train).unsqueeze(1)
X_test_t = torch.tensor(X_test)
Y_test_t = torch.tensor(Y_test).unsqueeze(1)

class SimpleClassifier(nn.Module):
    def __init__(self, input_dim):
        super(SimpleClassifier, self).__init__()
        self.layers = nn.Sequential(
            nn.Linear(input_dim, 16),
            nn.ReLU(),
            nn.Dropout(0.5),
            nn.Linear(16, 16),
            nn.ReLU(),
            nn.Dropout(0.5),
            nn.Linear(16, 1),
            nn.Sigmoid()
        )

    def forward(self, x):
        return self.layers(x)

model = SimpleClassifier(num_words)
criterion = nn.BCELoss()
optimizer = optim.Adam(model.parameters(), lr=0.001)
epochs = 5
batch_size = 512

print("Starting training...")
model.train()

for epoch in range(epochs):
    for i in range(0, len(X_train_t), batch_size):
        inputs =  X_train_t[i: i + batch_size]
        targets = Y_train_t[i: i + batch_size]
        optimizer.zero_grad()
        outputs = model(inputs)
        loss = criterion(outputs, targets)
        loss.backward()
        optimizer.step()

    print(f"Epoch {epoch + 1: 02d} completed | Loss: {loss.item():.4f}")

model.eval()
correct = 0
total = len(X_test_t)

with torch.no_grad():
    for i in range(0, total, batch_size):
        inputs = X_test_t[i: i + batch_size]
        targets = Y_test_t[i: i + batch_size]
        outputs = model(inputs)
        preds = (outputs > 0.5).float()
        correct += (preds == targets).sum().item()

accuracy = correct / total
print(f"Final test accuracy: {accuracy * 100:.2f}%")
```

```python
import tensorflow as tf
from tensorflow.keras.datasets import imdb
from tensorflow.keras.models import Sequential
from tensorflow.keras.layers import Dense,Dropout
import numpy as np
```

```python
tf.keras.utils.set_random_seed(42)
num_words = 10000

print("Loading IMDB dataset from keras...")
(X_train, Y_train), (X_test, Y_test) = imdb.load_data(num_words = num_words)

def vectorize_sequences(sequences, dimension = num_words):
    results = np.zeros((len(sequences), dimension), dtype = np.float32)
    for i, sequence in enumerate(sequences):
        results[i, sequence] = 1.0
    return results

print("Vectorizing training and testing data...")
X_train_vec = vectorize_sequences(X_train, num_words)
X_test_vec = vectorize_sequences(X_test, num_words)

Y_train = np.array(Y_train, dtype = np.float32)
Y_test = np.array(Y_test, dtype = np.float32)

model = Sequential([
    Dense(16, activation = "relu", input_shape = (num_words,)),
    Dropout(0.5),
    Dense(16, activation = "relu"),
    Dropout(0.5),
    Dense(1, activation = "sigmoid")
])

model.compile(
    optimizer = tf.keras.optimizers.Adam(learning_rate = 0.001),
    loss = "binary_crossentropy",
    metrics = ["accuracy"],
)

model.summary()
epochs = 5
batch_size = 512

history = model.fit(
    X_train_vec,
    Y_train,
    epochs = epochs,
    batch_size = batch_size,
    verbose = 1
)

test_loss, test_acc = model.evaluate(X_test_vec, Y_test, verbose = 0)
print(f"\nTest Loss: {test_loss:.4f} \nTest Accuracy: {test_acc * 100:.2f}%")

```
