```python
import pandas as pd

from sklearn.preprocessing import LabelEncoder
from sklearn.model_selection import train_test_split, cross_val_score
from sklearn.linear_model import Lasso, Ridge
from sklearn.metrics import mean_squared_error, r2_score
```

```python
df = pd.read_csv('Housing.csv')
df.head()
```

```python
for col in df.columns:
    if df[col].dtype == 'object':
        df[col] = LabelEncoder().fit_transform(df[col])

Xl = df.drop('price', axis = 1)
Yl = df['price']

Xl_train, Xl_test, Yl_train, Yl_test = train_test_split(
    Xl,
    Yl,
    test_size = 0.2,
    random_state = 42
)
model_lasso = Lasso()
model_lasso.fit(Xl_train, Yl_train)
Yl_pred = model_lasso.predict(Xl_test)
print(f'\nMean Squared Error: {mean_squared_error(Yl_test, Yl_pred)}')
print(f'\nR2 Score: {r2_score(Yl_test, Yl_pred)}')
lasso_scores = cross_val_score(
    model_lasso,
    Xl,
    Yl,
    cv = 10,
    scoring = 'r2'
)
print(f'\nCross Validation Score: {lasso_scores}')
print(f'\nAverage Score: {lasso_scores.mean()}')
```

```python
for col in df.columns:
    if df[col].dtype == 'object':
        df[col] = LabelEncoder().fit_transform(df[col])

Xr = df.drop('price', axis = 1)
Yr = df['price']

Xr_train, Xr_test, Yr_train, Yr_test = train_test_split(
    Xr,
    Yr,
    test_size = 0.2,
    random_state = 42
)
model_ridge = Ridge()
model_ridge.fit(Xr_train, Yr_train)
Yr_pred = model_ridge.predict(Xr_test)
print(f'\nMean Squared Error: {mean_squared_error(Yr_test, Yr_pred)}')
print(f'\nR2 Score: {r2_score(Yr_test, Yr_pred)}')
ridge_scores = cross_val_score(
    model_ridge,
    Xr,
    Yr,
    cv = 10,
    scoring = 'r2'
)
print(f'\nCross Validation Score: {ridge_scores}')
print(f'\nAverage Score: {ridge_scores.mean()}')
```
